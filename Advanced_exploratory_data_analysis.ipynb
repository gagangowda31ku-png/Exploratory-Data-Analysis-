```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("darkgrid")
iris_df = sns.load_dataset('iris')
tips_df = sns.load_dataset('tips')
```

```python
iris_df.columns
```

```python
num_col = iris_df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
```

```python
sns.barplot(data = iris_df , x = 'sepal_length' , y = 'species')
```

```python
g = sns.JointGrid(data = iris_df , x = 'sepal_length' , y = 'sepal_width')
g.plot(sns.scatterplot , sns.histplot)             
```

```python
g = sns.JointGrid(data = iris_df , x = 'petal_length' , y = 'petal_width')
g.plot(sns.kdeplot , sns.histplot)             
```

```python
g = sns.JointGrid(data = iris_df , x = 'petal_length' , y = 'petal_width' ,hue = 'species', palette = 'Set2')
g.plot(sns.scatterplot ,sns.kdeplot)         
```

```python
g = sns.PairGrid(iris_df)
g.map(sns.scatterplot)
```

```python
g = sns.PairGrid(iris_df , hue = 'species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
g.add_legend
```

```python
g = sns.PairGrid(iris_df , hue = 'species',palette = 'Set2')
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
g.add_legend
```

```python
sns.scatterplot(data =iris_df , x = 'sepal_length' , y = 'petal_length' , size = iris_df['sepal_length']*100 ,hue = 'petal_width')
plt.title("Bubblr plot of sepal_length and petal_length ")
```

```python
plt.scatter(iris_df['sepal_length'] , iris_df['sepal_width'] , s = iris_df['petal_length']*100 , c = iris_df['petal_width'])
plt.title("Bubblr plot")
plt.xlabel("sepal_length")
plt.ylabel("sepal_width")
```

```python
sns.stripplot(data = iris_df , y= 'sepal_length' , x = 'species' , jitter = True)
```

```python
sns.violinplot(data = iris_df , x= 'species' , y = 'sepal_length' , hue = 'species' , palette = "Set2")
```

```python
plt.hexbin(iris_df['sepal_length'] , iris_df['petal_length'], gridsize = 30 , cmap = 'plasma')
plt.colorbar(label = 'Number of observation')
plt.xlabel("---------")
plt.ylabel("============")
plt.show()
```

```python
sns.kdeplot(data = iris_df , x = 'sepal_length' , hue= 'species' , fill = True , common_norm = False)
```
